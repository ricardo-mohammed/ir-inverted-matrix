```python
# TF-IDF and Cosine Similarity Example
# ------------------------------------
# Goal: Show how to calculate TF-IDF and similarity between documents

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Our sample documents
docs = [
    "cat sat mat",
    "dog sat log",
    "cat chased dog"
]

# 2. Create a TF-IDF Vectorizer
# stop_words='english' removes common English words
vectorizer = TfidfVectorizer(stop_words='english')

# 3. Fit the model and transform the documents into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform(docs)

# 4. See what terms we have
print("Vocabulary:", vectorizer.get_feature_names_out())

# 5. View the TF-IDF matrix (as a dense table)
print("\nTF-IDF Matrix:\n", tfidf_matrix.toarray())

# 6. Compute pairwise cosine similarities
similarities = cosine_similarity(tfidf_matrix)

# 7. Show results
print("\nCosine Similarity Matrix:\n", similarities)
```
